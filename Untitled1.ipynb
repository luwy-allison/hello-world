```python
import pandas as pd
df = pd.read_csv("./data.csv")
```

```python
# get emotion dataframe
# column: definition
# 0: Answer (text): field with the respondent's answer to: "What were salient aspects of your day yesterday? How did you feel about them?”
# 1: Answer.f1.afraid.raw (boolean): whether the respondent labelled afraid as an emotion
# 2: Answer.f1.angry.raw (boolean): whether the respondent labelled angry as an emotion
# 3: Answer.f1.anxious.raw (boolean): whether the respondent labelled anxious as an emotion
# 4: Answer.f1.ashamed.raw (boolean): whether the respondent labelled ashamed as an emotion
# 5: Answer.f1.awkward.raw (boolean): whether the respondent labelled awkward as an emotion
# 6: Answer.f1.bored.raw (boolean): whether the respondent labelled bored as an emotion
# 7: Answer.f1.calm.raw (boolean): whether the respondent labelled calm as an emotion
# 8: Answer.f1.confused.raw (boolean): whether the respondent labelled confused as an emotion
# 9: Answer.f1.disgusted.raw (boolean): whether the respondent labelled disgusted as an emotion
# 10: Answer.f1.excited.raw (boolean): whether the respondent labelled excited as an emotion
# 11: Answer.f1.frustrated.raw (boolean): whether the respondent labelled frustrated as an emotion
# 12: Answer.f1.happy.raw (boolean): whether the respondent labelled happy as an emotion
# 13: nswer.f1.jealous.raw (boolean): whether the respondent labelled jealous as an emotion
# 14: Answer.f1.nostalgic.raw (boolean): whether the respondent labelled nostalgic as an emotion
# 15: Answer.f1.proud.raw (boolean): whether the respondent labelled proud as an emotion
# 16: Answer.f1.sad.raw (boolean): whether the respondent labelled sad as an emotion
# 17: Answer.f1.satisfied.raw (boolean): whether the respondent labelled satisfied as an emotion
# 18: Answer.f1.surprised.raw (boolean): whether the respondent labelled surprised as an emotion
label_to_test_dict = {
    16: 0, #sadness
    12: 1, #joy
    2: 3, #anger
    1: 4, #fear
    18: 5, #surprise
}

emotion_df = df.iloc[:, 0:19]

single_data_row = []
for index, row in emotion_df.iterrows():
    # print(index, row['Answer'])
    count = 0
    for column in emotion_df:
        if row[column] == True:
            count+=1
    if count == 1:
        single_data_row.append(index)

single_data_df = emotion_df.iloc[single_data_row,:] # filter data with single emotion
labels = []
for index, row in single_data_df.iterrows():
    # print(index, row['Answer'])
    col = 0
    for column in single_data_df:
        if row[column] == True:
            # print(column,col)
            # labels.append(column.split('.',)[2])
            labels.append(col)
            break
        col+=1

text_label = single_data_df.iloc[:,0:1]
text_label = text_label.assign(label = labels)
text_label

for index, row in text_label.iterrows():
    if row['label'] not in label_to_test_dict.keys():
      text_label.drop(index, inplace=True)

text_label['label'] = text_label['label'].map(label_to_test_dict)
text_label
```

```python
from sklearn.utils import resample

# Check the distribution of each class
print(CARER_text_label['label'].value_counts())

# Find the minimum class size
min_class_size = CARER_text_label['label'].value_counts().min()

# Create a DataFrame to store the undersampled data
undersampled_df = pd.DataFrame(columns=CARER_text_label.columns)

# Perform undersampling
for label in CARER_text_label['label'].unique():
    # Select all samples of this class
    class_samples = CARER_text_label[CARER_text_label['label'] == label]

    # Undersample to the size of the smallest class
    undersampled_samples = resample(class_samples,
                                    replace=False,  # Don't replace the samples
                                    n_samples=min_class_size,  # Match number of samples to smallest class
                                    random_state=42)  # Set seed for reproducibility

    # Append the undersampled data to the DataFrame
    undersampled_df = pd.concat([undersampled_df, undersampled_samples])

# Shuffle the undersampled dataset to mix different classes
undersampled_df = undersampled_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Check the new class distribution after undersampling
print(undersampled_df['label'].value_counts())
```

```python
from sklearn.utils import resample

# Step 1: Check the distribution of each class
print(CARER_text_label['label'].value_counts())

# Step 2: Perform random oversampling
# Create a list to store the resampled DataFrames
df_resampled = []

# Find the maximum number of samples in any class
max_samples = CARER_text_label['label'].value_counts().max()

# Loop through each unique label
for label in CARER_text_label['label'].unique():
    # Select all rows for the current label
    df_label = CARER_text_label[CARER_text_label['label'] == label]

    # If the current class has fewer samples than the maximum, perform oversampling
    df_label_oversampled = resample(df_label,
                                    replace=True,            # Allow resampling with replacement
                                    n_samples=max_samples,   # Number of samples to match the largest class
                                    random_state=42)         # Set a seed for reproducibility

    # Append the resampled DataFrame to the list
    df_resampled.append(df_label_oversampled)

# Step 3: Concatenate all resampled DataFrames into one
CARER_text_label_oversampled = pd.concat(df_resampled)

# Step 4: Check the distribution after oversampling to confirm balancing
print(CARER_text_label_oversampled['label'].value_counts())
```

```python
from transformers.models.bert.modeling_bert import BertPreTrainedModel, BertModel
import torch.nn as nn
import torch
from transformers import AutoTokenizer
import transformers

class BERTClassifier(nn.Module):

    def __init__(self, output_dim, pretrained_name='bert-base-uncased'):

        super(BERTClassifier, self).__init__()

        self.bert = BertModel.from_pretrained(pretrained_name)

        self.mlp = nn.Linear(768, output_dim)


    def forward(self, tokens_X):

        res = self.bert(**tokens_X)

        return self.mlp(res[1])
```

```python
def evaluate(net, text_data, labels_data):

    sum_correct, i = 0, 0

    while i < len(text_data):

        # comments = text_data[i: min(i + 10, len(text_data))]
        target = text_data[i]

        tokens_X = tokenizer(target, padding=True, truncation=True, return_tensors='pt').to(device=device)

        res = net(tokens_X)

        # y = torch.tensor(labels_data[i: min(i + 10, len(text_data))]).reshape(-1).to(device=device)
        y = torch.tensor(labels_data[i]).reshape(-1).to(device=device)
        sum_correct += (res.argmax(axis=1) == y).sum()
        # i += 10
        i += 1

    return sum_correct/len(text_data)
```

```python
def train_bert_classifier(model, tokenizer, loss, optimizer, train_text, train_labels, test_text, test_labels, device, epochs):

    max_acc = 0.5
    model.to(device)

    for epoch in range(epochs):

        # i, sum_loss = 0, 0
        i = 0
        # train_acc = evaluate(model, train_text, train_labels)
        # test_acc = evaluate(model, test_text, test_labels)
        sum_loss = 0
        # print('\n--epoch', epoch, '\t--loss:', sum_loss / len(train_text), '\t--train_acc:', train_acc, '\t--test_acc', test_acc)

        while i < len(train_text):
            # comments = train_comments[i: min(i+10, len(train_comments))]
            train_x = train_text[i: min(i+10, len(train_text))]
            # print(i)

            tokens_X = tokenizer(train_x, padding=True, truncation=True, return_tensors='pt').to(device=device)


            res = model(tokens_X)


            y = torch.tensor(train_labels[i: min(i+10, len(train_labels))]).reshape(-1).to(device=device)
            # y = torch.tensor(train_labels[i]).reshape(-1).to(device=device)

            optimizer.zero_grad()
            l = loss(res, y)
            l.backward()
            optimizer.step()

            sum_loss += l.detach()
            i += 10
            # i+=1

        train_acc = evaluate(model, train_text, train_labels)
        test_acc = evaluate(model, test_text, test_labels)

        print('\n--epoch', epoch, '\t--loss:', sum_loss / len(train_text)/10, '\t--train_acc:', train_acc, '\t--test_acc', test_acc)

        if test_acc > max_acc:
          max_acc = test_acc
          torch.save(model.state_dict(), 'bert.parameters')
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = BERTClassifier(output_dim = 6)
config_name = 'bert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(config_name)
loss_fct = torch.nn.CrossEntropyLoss() # we use cross entrophy loss

# optimizer = torch.optim.Adam(model.parameters(), lr=parameters['learning_rate'], betas=(0.9, 0.999), eps=1e-9)
optimizer = torch.optim.SGD(model.parameters(), lr=1e-4)
```

```python
import torch
from transformers import AutoTokenizer
model_path = 'bert.parameters'
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
config_name = 'bert-base-uncased'
tokenizer = AutoTokenizer.from_pretrained(config_name)
model = torch.load(model_path)
label_dict = {
    "0": "sadness",
    "1": "joy",
    "2": "love",
    "3": "anger",
    "4": "fear",
    "5": "surprise",
}
def predict(text):
    print(text,"\n")
    pred = model(tokenizer(text, padding=True, truncation=True, return_tensors='pt').to(device=device))
    result = label_dict[str(pred.argmax(axis = 1).item())]
    print(pred.argmax(axis = 1).item(), result)

    return result
```
